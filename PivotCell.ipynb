```python

import pandas as pd
from google.colab import files
from IPython.display import display

# ==========================================
# CARGAR LOS DOS ARCHIVOS
# ==========================================

print("Sube adult.csv y antropometria-dataset-2.csv")
files.upload()

peso = pd.read_csv("antropometria-dataset-2.csv")
adult = pd.read_csv("adult.csv")

# Limpiar espacios
peso["desc_ent"] = peso["desc_ent"].str.strip()

for col in ["sex", "education", "marital-status"]:
    adult[col] = adult[col].str.strip()

# ==========================================
# 1. PROMEDIO DE PESO POR ESTADO
# ==========================================

print("\n1. AVERAGE WEIGHT BY STATE")

tabla1 = pd.pivot_table(
    peso,
    index="desc_ent",
    values="peso",
    aggfunc="mean"
)

tabla1.columns = ["Average Weight"]
display(tabla1.style.format("{:.4f}"))

# ==========================================
# 2 Y 3. PESO PROMEDIO POR SEXO Y ESTADO
# ==========================================

print("\n2 Y 3. AVERAGE WEIGHT BY STATE AND SEX")

tabla2 = pd.pivot_table(
    peso,
    index="desc_ent",
    columns="sexo",
    values="peso",
    aggfunc="mean",
    margins=True,
    margins_name="All"
)

tabla2 = tabla2.rename(columns={
    1: "Sex 1",
    2: "Sex 2"
})

display(tabla2.style.format("{:.4f}"))

# ==========================================
# 4. DOS PARRAFOS DE INTERPRETACION
# ==========================================

print("\n4. INTERPRETATION OF RESULTS")

promedios = peso.groupby("desc_ent")["peso"].mean()

mayor = promedios.idxmax()
menor = promedios.idxmin()
promedio_general = peso["peso"].mean()

promedio_sexo = peso.groupby("sexo")["peso"].mean()

print(
    f"\nParagraph 1: The data shows differences in "
    f"average weight among Mexican states. "
    f"{mayor} has the highest average weight "
    f"with {promedios[mayor]:.4f} kg, while "
    f"{menor} has the lowest average weight "
    f"with {promedios[menor]:.4f} kg. "
    f"The overall average weight is "
    f"{promedio_general:.4f} kg. "
    f"This shows that average weight varies "
    f"depending on the state analyzed."
)

print(
    f"\nParagraph 2: When comparing the two sex "
    f"categories, sex 1 has an overall average "
    f"weight of {promedio_sexo[1]:.4f} kg, "
    f"while sex 2 has an average of "
    f"{promedio_sexo[2]:.4f} kg. "
    f"The difference between these averages "
    f"is {abs(promedio_sexo[1]-promedio_sexo[2]):.4f} kg. "
    f"These results demonstrate differences "
    f"between the groups in the dataset, "
    f"although they do not explain the causes."
)

# ==========================================
# 5. NUMERO DE PERSONAS POR ESTADO CIVIL
# ==========================================

print("\n5. NUMBER OF PEOPLE BY MARITAL STATUS")

tabla3 = pd.pivot_table(
    adult,
    index="marital-status",
    values="age",
    aggfunc="count"
)

tabla3.columns = ["Number of People"]

display(tabla3)

# ==========================================
# 6. PORCENTAJE POR ESTADO CIVIL
# ==========================================

print("\n6. MARITAL STATUS AND PERCENTAGE")

tabla3["Percentage (%)"] = (
    tabla3["Number of People"] /
    tabla3["Number of People"].sum() * 100
)

tabla3 = tabla3.sort_values(
    "Number of People",
    ascending=False
)

display(tabla3.style.format({
    "Number of People": "{:.0f}",
    "Percentage (%)": "{:.4f}"
}))

print("\nTotal people:", tabla3["Number of People"].sum())
print("Total percentage:",
      f"{tabla3['Percentage (%)'].sum():.4f}%")

# ==========================================
# PREGUNTAS ANTERIORES DE LA ACTIVIDAD
# ==========================================

print("\n========== ADDITIONAL QUESTIONS ==========")

# Average weight in Sinaloa
sinaloa = peso.loc[
    peso["desc_ent"] == "SINALOA", "peso"
].mean()

print(f"\n1. Average weight SINALOA: {sinaloa:.4f}")

# Median weight sex 1 in Jalisco
jalisco = peso.loc[
    (peso["desc_ent"] == "JALISCO") &
    (peso["sexo"] == 1), "peso"
].median()

print(f"2. Median weight sex 1 JALISCO: {jalisco:.4f}")

# Minimum average weight for sex 2
sexo2 = peso[peso["sexo"] == 2].groupby(
    "desc_ent"
)["peso"].mean()

estado_min = sexo2.idxmin()

print(
    f"3. Minimum average sex 2: "
    f"{estado_min} = {sexo2.min():.4f}"
)

# Average age of Divorced with HS-grad
hs = adult.loc[
    (adult["marital-status"] == "Divorced") &
    (adult["education"] == "HS-grad"),
    "age"
].mean()

print(f"4. Average age Divorced HS-grad: {hs:.4f}")

# Median age Female Divorced 12th
mediana12 = adult.loc[
    (adult["sex"] == "Female") &
    (adult["marital-status"] == "Divorced") &
    (adult["education"] == "12th"),
    "age"
].median()

print(
    f"5. Median age Female Divorced 12th: "
    f"{mediana12:.4f}"
)

# ==========================================
# DESCARGAR TABLAS EN CSV
# ==========================================

tabla1.to_csv("average_weight_by_state.csv")
tabla2.to_csv("average_weight_by_sex_and_state.csv")
tabla3.to_csv("marital_status_percentages.csv")

print("\nAll calculations completed successfully.")
print("CSV tables have been generated.")
```

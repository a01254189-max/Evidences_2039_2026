```python
import io
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
from google.colab import files

# 1. Sube el archivo antropometria-dataset-1-1.csv
archivos = files.upload()
nombre_csv = next(nombre for nombre in archivos if nombre.lower().endswith('.csv'))
df = pd.read_csv(io.BytesIO(archivos[nombre_csv]))

sns.set_theme(style='darkgrid')
print('Datos cargados:', len(df), 'registros')

# Pregunta 1: Box plot de EDAD para toda la base -> opción 2
plt.figure(figsize=(9, 3))
sns.boxplot(x=df['edad'], whis=(0, 100), color='steelblue')
plt.title('P1. Edad total — Respuesta: 2')
plt.xlabel('Edad (años)')
plt.tight_layout()
plt.show()

# Pregunta 2: Mediana de EDAD por estado para SEXO 2 -> opción B
edad_sexo2 = (df[df['sexo'] == 2]
              .groupby('desc_ent')['edad'].median().sort_index())
plt.figure(figsize=(9, 3))
sns.boxplot(x=edad_sexo2, whis=(0, 100), color='steelblue')
plt.title('P2. Mediana de edad por estado, sexo 2 — Respuesta: B')
plt.xlabel('Mediana de edad (años)')
plt.tight_layout()
plt.show()

# Pregunta 3: Line plot de mediana de CINTURA por estado para SEXO 1 -> opción C
cintura_sexo1 = (df[df['sexo'] == 1]
                 .groupby('desc_ent')['cintura'].median().sort_index())
plt.figure(figsize=(9, 3))
plt.plot(range(len(cintura_sexo1)), cintura_sexo1.values)
plt.title('P3. Mediana de cintura por estado, sexo 1 — Respuesta: C')
plt.xlabel('Índice de estado (orden alfabético)')
plt.ylabel('Cintura (cm)')
plt.tight_layout()
plt.show()

# Pregunta 4: Las 10 mayores SUMAS de SISTOL por estado para SEXO 1 -> opción B
sistol_sexo1 = (df[df['sexo'] == 1]
                .groupby('desc_ent')['sistol'].sum()
                .nlargest(10).sort_values())
plt.figure(figsize=(11, 5))
plt.barh(sistol_sexo1.index, sistol_sexo1.values, color='slateblue')
plt.title('P4. Top 10 de suma de sistol, sexo 1 — Respuesta: B')
plt.xlabel('Suma de sistol')
plt.tight_layout()
plt.show()

# Pregunta 5: Histograma de PESO con 25 intervalos -> opción C
plt.figure(figsize=(9, 4))
plt.hist(df['peso'].dropna(), bins=25, color='steelblue')
plt.title('P5. Histograma de peso (bins=25) — Respuesta: C')
plt.xlabel('Peso (kg)')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()

# Tabla de referencia, igual a la que aparece en las preguntas
print('Mediana del peso por estado y sexo (primeros tres estados):')
mediana_peso = pd.pivot_table(df, values='peso', index='desc_ent',
                             columns='sexo', aggfunc='median', margins=True)
display(mediana_peso.head(3))
print('\nRESPUESTAS: 1 = 2 | 2 = B | 3 = C | 4 = B | 5 = C')
```
